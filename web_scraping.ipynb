```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import re
import time

all_data = []

for page in range(1, 200):
    url = f"https://sheffieldegypt.com/collections/all-products?page={page}"
    
    response = requests.get(url)
    soup = BeautifulSoup(response.text, "html.parser")
    
    product_links = soup.select('h3 a[href^="/products/"]')
    
    for a in product_links:
        name = a.get_text(strip=True)
        link = "https://sheffieldegypt.com" + a["href"]
        
        block = a.find_parent("div", class_="sf__pcard")
        text = block.get_text(" ", strip=True) if block else ""
        
        prices = re.findall(r"LE\s?[\d,]+\.\d{2}", text)
        
        current_price = prices[0] if len(prices) >= 1 else None
        old_price = prices[1] if len(prices) >= 2 else None
        
        all_data.append({
            "Name": name,
            "Current Price": current_price,
            "Old Price": old_price,
            "URL": link
        })
        
        if len(all_data) >= 1000:
            break
    
    print(f"Page {page} done | Total rows = {len(all_data)}")
    
    if len(all_data) >= 1000:
        break
    
    time.sleep(0.5)

df = pd.DataFrame(all_data)

print(df.shape)
df.head()
```

```python
df.to_csv("sheffield_raw_data.csv", index=False)
```
